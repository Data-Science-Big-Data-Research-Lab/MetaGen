# The problem

The Rastrigin function in two dimensions has a regular grid of local minima around its global minimum at the origin, where it is worth 0.

Domain:
$$x, y \models Def^{R} = \langle -5.12, 5.12\rangle$$

Fitness function:
$$20 + x^2 - 10\cos(2\pi x) + y^2 - 10\cos(2\pi y)$$

```python
import math

from metagen.framework import Domain, Solution
from metagen.metaheuristics import TabuSearch
```

## Domain and fitness function

```python
domain: Domain = Domain()
domain.define_real("x", -5.12, 5.12)
domain.define_real("y", -5.12, 5.12)


def rastrigin(solution: Solution) -> float:
    x, y = solution["x"], solution["y"]
    return 20 + x ** 2 - 10 * math.cos(2 * math.pi * x) + y ** 2 - 10 * math.cos(2 * math.pi * y)
```

## Running the tabu search

A single current solution walks the search space. In each iteration it samples `population_size` neighbors and moves to the best one that is not tabu, even when it is worse than the current solution, which is what lets the walk leave a local minimum. The last `tabu_size` visited solutions are tabu, and `tabu_radius` says how close to one of them a neighbor has to be to count as visited. The `seed` makes the run reproducible.

```python
tabu_search = TabuSearch(domain, rastrigin, population_size=10, max_iterations=50, tabu_size=10, seed=0)
best_solution: Solution = tabu_search.run()

print(best_solution)
```

## Reading the result

The solution returned is the best one found along the walk, not the point where the walk ended. Its variables are read like a dictionary, and the history of the best fitness per iteration is kept in the algorithm.

```python
print("x =", best_solution["x"])
print("y =", best_solution["y"])
print("fitness =", best_solution.get_fitness())
print("best fitness per iteration:", [round(f, 3) for f in tabu_search.best_solution_fitnesses])
```
